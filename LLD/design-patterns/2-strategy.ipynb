## Strategy Design Pattern:

Strategy design pattern is a behavioural pattern that defines a family of algorithms, encapsulate each one in a separate class and make them interchangable at run time.

It enables selecting an algorithm's implementation independently of the client (context) using it, promoting code flexibility, maintainability, and adherence to the Open/Closed Principle

### When to Use It:
1. You need to switch algorithms at runtime (e.g., changing payment methods like Credit Card, PayPal, or Bitcoin).
2. You have multiple, similar conditional statements (if-else/switch) selecting different behaviors.
3. You want to separate business logic from implementation details. 

### Real-World Examples:
1. Multiple payment methods (Credit Card, PayPal, Bank Transfer, UPI etc.) on an E-Commerce site. Each payment method has its own way to process, verify, handle failures, possibly different fees. Defining a PaymentStrategy interface and concrete implementations for each payment method picks the right one based on user choice at runtime.
2. Shipping Cost Calculation: Calculating shipping via UPS, FedEx, or postal service.
3. Sorting Algorithms: Switching between algorithms like quicksort or bubblesort based on input data.
4. File Compression: Choosing between ZIP, RAR, or TAR compression.

#### Advantages:
1. Improved Maintainability: Strategies can be added or modified without changing the context.
2. Reduced Conditional Code: Eliminates complex if-else or switch statements.
3. Encapsulation: Hides implementation details from the client. 

#### Disadvantages:
1. Increased Object Count: Introduces new classes for each strategy.
2. Client Awareness: The client must understand the differences between strategies to select the correct one.

```python
from abc import ABC, abstractmethod

class SortingStrategy(ABC):
    @abstractmethod
    def sort(self, array):
        pass

class BubbleSortStrategy(SortingStrategy):
    def sort(self, array):
        print('Sorting using Bubble Sort')
        # Bubble sort implementation

class MergeSortStrategy(SortingStrategy):
    def sort(self, array):
        print('Sorting using Merge Sort')
        # Merge sort implementation

class QuickSortStrategy(SortingStrategy):
    def sort(self, array):
        print('Sorting using Quick Sort')
        # Quick sort implementation

class SortingContext:
    def __init__(self, strategy):
        self.strategy = strategy

    def perform_sort(self, array):
        self.strategy.sort(array)

    def set_sorting_strategy(self, strategy):
        self.strategy = strategy

# Create SortingContext with BubbleSortStrategy
sorting_context = SortingContext(BubbleSortStrategy())
array1 = ['cpp','c','java','python3','csharp','html','css','javascript','php','cpp14','cobol','dart','go','julia','kotlin','lisp','matlab','node','objc','perl','r','rust','ruby','scala','swift','solidity','xml']
sorting_context.perform_sort(array1) # Output: Sorting using Bubble Sort

# Change strategy to MergeSortStrategy
sorting_context.set_sorting_strategy(MergeSortStrategy())
array2 = ['cpp','c','java','python3','csharp','html','css','javascript','php','cpp14','cobol','dart','go','julia','kotlin','lisp','matlab','node','objc','perl','r','rust','ruby','scala','swift','solidity','xml']
sorting_context.perform_sort(array2) # Output: Sorting using Merge Sort

# Change strategy to QuickSortStrategy
sorting_context.set_sorting_strategy(QuickSortStrategy())
array3 = ['cpp','c','java','python3','csharp','html','css','javascript','php','cpp14','cobol','dart','go','julia','kotlin','lisp','matlab','node','objc','perl','r','rust','ruby','scala','swift','solidity','xml']
sorting_context.perform_sort(array3) # Output: Sorting using Quick Sort
```

```python
from abc import ABC, abstractmethod

class WalkableRobot(ABC):
    @abstractmethod
    def walk(self):
        pass

class NormalWalkableRobot(WalkableRobot):
    def walk(self):
        print("Normal walkable robot")

class NoWalkableRobot(WalkableRobot):
    def walk(self):
        print("No walkable robot")

class TalkableRobot(ABC):
    @abstractmethod
    def talk(self):
        pass

class NormalTalkableRobot(TalkableRobot):
    def talk(self):
        print("Normal talk robot")

class NoTalkableRobot(TalkableRobot):
    def talk(self):
        print("No talk robot")

class FlyableRobot(ABC):
    @abstractmethod
    def fly(self):
        pass

class NormalFlyableRobot(FlyableRobot):
    def fly(self):
        print("Normal flyable robot")

class NoFlyableRobot(FlyableRobot):
    def fly(self):
        print("No flyable robot")

class RobotContext:
    def __init__(self, talkable_robot, walkable_robot, flyable_robot):
        self.talkable_robot = talkable_robot
        self.walkable_robot = walkable_robot
        self.flyable_robot = flyable_robot

    def talk(self):
        self.talkable_robot.talk()
    def walk(self):
        self.walkable_robot.walk()
    def fly(self):
        self.flyable_robot.fly()
    @abstractmethod
    def projection(self):
        pass

class CompanionRobot(RobotContext):
    
    def __init__(self, talkable_robot, walkable_robot, flyable_robot):
        super().__init__(talkable_robot, walkable_robot, flyable_robot)

    def projection(self):
        self.talkable_robot.talk()
        self.walkable_robot.walk()
        self.flyable_robot.fly()

        
class WorkerRobot(RobotContext):
    def __init__(self, talkable_robot, walkable_robot, flyable_robot):
        super().__init__(talkable_robot, walkable_robot, flyable_robot)
    def projection(self):
        self.talkable_robot.talk()
        self.walkable_robot.walk()
        self.flyable_robot.fly()

# Runtime
robot = CompanionRobot(NormalTalkableRobot(), NoWalkableRobot(), NormalFlyableRobot())
robot.projection()
print("--------------------------------")
robot = WorkerRobot(NormalTalkableRobot(), NormalWalkableRobot(), NoFlyableRobot())
robot.projection()
```
