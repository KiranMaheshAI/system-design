## Factory Design Pattern

Factory Method is a creational design pattern that provides an interface for creating objects in a superclass, but allows subclasses to alter the type of objects that will be created.

Factory Design pattern's main agenda is creating an object based on client request. If the request change, object will change.

Use the Factory Method when you don’t know beforehand the exact types and dependencies of the objects your code should work with.

The Factory Method separates product construction code from the code that actually uses the product. Therefore it’s easier to extend the product construction code independently from the rest of the code.

"Use the Factory Method when you want to save system resources by reusing existing objects instead of rebuilding them each time."

There are three types of factory Design patterns:
1. Simple Factory
2. Factory Method
3. Abstract Factory Method

### Advantages
The Factory Method pattern provides flexibility and better design structure in object creation:

1. Promotes loose coupling by separating object creation from its usage.
2. Makes the code open for extension and closed for modification (supports OCP).
3. Improves maintainability by allowing new product types to be added easily.

### Disadvantages
Although useful, the pattern can introduce some complexity:

1. Increases the number of classes in the system.
2. Can make the code more complex compared to simple object creation.
3. May be unnecessary for small or simple applications.

### Simple Factory

A factory class decides which concrete class to initialize.

```python
from abc import ABC, abstractmethod

class Burger(ABC):
    @abstractmethod
    def prepare(self):
        pass

class BasicBurger(Burger):
    def prepare(self):
        print("Preparing Basic Burger")

class StandardBurger(Burger):
    def prepare(self):
        print("Preparing Standard Burger")

class DeluxeBurger(Burger):
    def prepare(self):
        print("Preparing Deluxe Burger")
        
class BurgerFactory:
    def __init__(self, burger_type):
        self.burger_type = burger_type

    def get_burger(self):
        if self.burger_type == "basic":
            return BasicBurger()
        elif self.burger_type == "standard":
            return StandardBurger()
        elif self.burger_type == "deluxe":
            return DeluxeBurger()
        else:
            raise ValueError("Invalid burger type")

def main():
    burger_factory = BurgerFactory("basic")
    burger = burger_factory.get_burger()
    burger.prepare()

main()
```

### Factory Method: 

Factory method design pattern defines an interface for creating objects which allows subclasses to decide which class to instantiate.

```python
class Burger(ABC):
    @abstractmethod
    def prepare(self):
        pass

class BasicBurger(Burger):
    def prepare(self):
        print("Preparing Basic Burger")

class StandardBurger(Burger):
    def prepare(self):
        print("Preparing Standard Burger")

class PremiumBurger(Burger):
    def prepare(self):
        print("Preparing Premium Burger")

class BasicWheatBurger(Burger):
    def prepare(self):
        print("Preparing Basic Wheat Burger")

class StandardWheatBurger(Burger):
    def prepare(self):
        print("Preparing standard wheat Burger")

class PremiumWheatBurger(Burger):
    def prepare(self):
        print("Preparing premium wheat Burger")

class BurgerFactory(ABC):
    @abstractmethod
    def create_burger(self, type):
        pass

class SinghBurgerFactory(BurgerFactory):
    def create_burger(self, type):
        if type == "basic":
            return BasicBurger()
        elif type == "standard":
            return StandardBurger()
        elif type == "premium":
            return PremiumBurger()
        else:
            raise ValueError("Invalid burger type")

class KinnaryBurgerFactory(BurgerFactory):
    def create_burger(self, type):
        if type == "basic":
            return BasicWheatBurger()
        elif type == "standard":
            return StandardWheatBurger()
        elif type == "premium":
            return PremiumWheatBurger()
        else:
            raise ValueError("Invalid burger type")

def main():
    type = "basic"
    burger_factory = SinghBurgerFactory()
    burger = burger_factory.create_burger(type)
    burger.prepare()

main()
        
```

```python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def printVehicle(self):
        pass

class TwoWheeler(Vehicle) {
    def printVehicle():
        print("I am two wheeler")
}

class FourWheeler(Vehicle):
    def printVehicle(self):
        print("I am four wheeler")

class VehicleFactory(ABC):
    @abstractmethod
    def createVehicle(self):
        pass


class TwoWheelerFactory(VehicleFactory):
    def createVehicle(self):
        return TwoWheeler()

class FourWheelerFactory(VehicleFactory):
    def createVehicle(self):
        return FourWheeler()

class Client:
    def __init__(self, factory):
        self.pVehicle = factory.createVehicle();

    def getVehicle(self):
        return self.pVehicle

def main():
    twoWheelerFactory = TwoWheelerFactory();
    twoWheelerClient = Client(twoWheelerFactory);
    twoWheeler = twoWheelerClient.getVehicle();
    twoWheeler.printVehicle();

    fourWheelerFactory = FourWheelerFactory();
    fourWheelerClient = Client(fourWheelerFactory);
    fourWheeler = fourWheelerClient.getVehicle();
    fourWheeler.printVehicle();

main();
```

### Abstract Factory Method

Abstract factory provides interface of creating families of related objects without specifying their concrete class

Use the Abstract Factory when your code needs to work with various families of related products, but you don’t want it to depend on the concrete classes of those products—they might be unknown beforehand or you simply want to allow for future extensibility.

The Abstract Factory provides you with an interface for creating objects from each class of the product family. As long as your code creates objects via this interface, you don’t have to worry about creating the wrong variant of a product which doesn’t match the products already created by your app.

Consider implementing the Abstract Factory when you have a class with a set of Factory Methods that blur its primary responsibility.

In a well-designed program each class is responsible only for one thing. When a class deals with multiple product types, it may be worth extracting its factory methods into a stand-alone factory class or a full-blown Abstract Factory implementation.

```python
class Burger(ABC):
    @abstractmethod
    def prepare(self):
        pass

class BasicBurger(Burger):
    def prepare(self):
        print("Preparing Basic Burger")

class StandardBurger(Burger):
    def prepare(self):
        print("Preparing Standard Burger")

class PremiumBurger(Burger):
    def prepare(self):
        print("Preparing Premium Burger")

class BasicWheatBurger(Burger):
    def prepare(self):
        print("Preparing Basic Wheat Burger")

class StandardWheatBurger(Burger):
    def prepare(self):
        print("Preparing standard wheat Burger")

class PremiumWheatBurger(Burger):
    def prepare(self):
        print("Preparing premium wheat Burger")

class GarlicBread(ABC):
    @abstractmethod
    def prepare(self):
        pass

class BasicGarlicBread(GarlicBread):
    def prepare(self):
        print("Preparing Basic Garlic Bread")

class CheeseGarlicBread(GarlicBread):
    def prepare(self):
        print("Preparing Standard Garlic Bread")

class BasicWheatGarlicBread(GarlicBread):
    def prepare(self):
        print("Preparing Basic Wheat Garlic Bread")

class CheeseWheatGarlicBread(GarlicBread):
    def prepare(self):
        print("Preparing Cheese Wheat Garlic Bread")


class MealFactory(ABC):
    @abstractmethod
    def createBurger(self, type):
        pass

    @abstractmethod
    def createGarlicBread(self, type):
        pass

class SinghMealFactory(MealFactory):
    def createBurger(self, type):
        if type == "basic": 
            return BasicBurger()
        elif type == "standard":
            return StandardBurger()
        elif type == "premium":
            return PremiumBurger()
        else:
            raise ValueError("Invalid burger type")

    def createGarlicBread(self, type):
        if type == "basic":
            return BasicGarlicBread()
        elif type == "cheese":
            return CheeseGarlicBread()

class KingMealFactory(MealFactory):
    def createBurger(self, type):
        if type == "basic":
            return BasicWheatBurger()
        elif type == "standard":
            return StandardWheatBurger()
        elif type == "premium":
            return PremiumWheatBurger()
        else:
            raise ValueError("Invalid burger type")

    def createGarlicBread(self, type):
        if type == "basic":
            return BasicWheatGarlicBread()
        elif type == "cheese":
            return CheeseWheatGarlicBread()
        else:
            raise ValueError("Invalid garlic bread type")

def main():
    burgerType = "basic"
    garlicBreadType = "cheese"
    mealFactory = SinghMealFactory()
    meal = mealFactory.createBurger(burgerType)
    meal = mealFactory.createGarlicBread(garlicBreadType)
    meal.prepare()
    meal.prepare()

main()
```

```python
"""MCP Authentication Strategies - Create httpx.Auth objects from credentials."""

import base64
from abc import ABC, abstractmethod
from enum import Enum
from typing import Dict, Optional, Type

import httpx

from core.security.mcp_credentials import MCPCredentials


class AuthType(Enum):
    """Supported authentication types for MCP servers."""

    NONE = "none"
    BASIC = "basic"
    BEARER = "bearer"
    API_KEY = "api_key"
    CUSTOM_HEADER = "custom_header"
    OAUTH2_CLIENT_CREDENTIALS = "oauth2_client_credentials"


# --- Interface ---
class IMCPAuthStrategy(ABC):
    """Interface for MCP authentication strategies that create httpx.Auth objects."""

    @abstractmethod
    async def create_auth(self, credentials: MCPCredentials) -> Optional[httpx.Auth]:
        """Create an httpx.Auth object from credentials.

        Args:
            credentials: MCPCredentials containing auth data

        Returns:
            httpx.Auth object or None if no auth needed
        """
        pass

    @abstractmethod
    def get_auth_headers(self, credentials: MCPCredentials) -> Dict[str, str]:
        """Get authentication headers from credentials.

        Args:
            credentials: MCPCredentials containing auth data

        Returns:
            Dictionary of HTTP headers for authentication
        """
        pass


# --- Concrete Strategies ---


class NoAuthStrategy(IMCPAuthStrategy):
    """No authentication strategy."""

    async def create_auth(self, credentials: MCPCredentials) -> Optional[httpx.Auth]:
        """No authentication needed."""
        return None

    def get_auth_headers(self, credentials: MCPCredentials) -> Dict[str, str]:
        """No headers needed."""
        return {}


class BasicAuthStrategy(IMCPAuthStrategy):
    """Basic authentication using username/password."""

    async def create_auth(self, credentials: MCPCredentials) -> Optional[httpx.Auth]:
        """Create httpx.BasicAuth from credentials."""
        if not credentials.username or not credentials.password:
            raise ValueError("Username and password required for basic auth")

        # httpx.BasicAuth is synchronous, but we wrap it in async to satisfy interface
        return httpx.BasicAuth(
            username=credentials.username, password=credentials.password
        )

    def get_auth_headers(self, credentials: MCPCredentials) -> Dict[str, str]:
        """Generate Basic Auth headers."""
        if not credentials.username or not credentials.password:
            raise ValueError("Username and password required for basic auth")

        auth_string = f"{credentials.username}:{credentials.password}"
        encoded = base64.b64encode(auth_string.encode()).decode()
        return {"Authorization": f"Basic {encoded}"}


# --- Placeholders for Future Implementation ---


class BearerAuthStrategy(IMCPAuthStrategy):
    """Bearer token authentication (Placeholder)."""

    async def create_auth(self, credentials: MCPCredentials) -> Optional[httpx.Auth]:
        raise NotImplementedError("Bearer auth not yet implemented")

    def get_auth_headers(self, credentials: MCPCredentials) -> Dict[str, str]:
        raise NotImplementedError("Bearer auth not yet implemented")


class APIKeyAuthStrategy(IMCPAuthStrategy):
    """API key authentication (Placeholder)."""

    async def create_auth(self, credentials: MCPCredentials) -> Optional[httpx.Auth]:
        raise NotImplementedError("API Key auth not yet implemented")

    def get_auth_headers(self, credentials: MCPCredentials) -> Dict[str, str]:
        raise NotImplementedError("API Key auth not yet implemented")


class CustomHeaderAuthStrategy(IMCPAuthStrategy):
    """Custom header authentication (Placeholder)."""

    async def create_auth(self, credentials: MCPCredentials) -> Optional[httpx.Auth]:
        raise NotImplementedError("Custom Header auth not yet implemented")

    def get_auth_headers(self, credentials: MCPCredentials) -> Dict[str, str]:
        raise NotImplementedError("Custom Header auth not yet implemented")


class OAuth2ClientCredentialsAuthStrategy(IMCPAuthStrategy):
    """OAuth2 Client Credentials flow (Placeholder)."""

    async def create_auth(self, credentials: MCPCredentials) -> Optional[httpx.Auth]:
        raise NotImplementedError("OAuth2 Client Credentials not yet implemented")

    def get_auth_headers(self, credentials: MCPCredentials) -> Dict[str, str]:
        raise NotImplementedError("OAuth2 Client Credentials not yet implemented")


# --- Factory ---
class AuthFactory:
    """Factory for creating MCP authentication strategies."""

    # Map auth types to strategy classes
    _strategy_map: Dict[str, Type[IMCPAuthStrategy]] = {
        AuthType.NONE.value: NoAuthStrategy,
        AuthType.BASIC.value: BasicAuthStrategy,
        # Registered but not yet implemented
        AuthType.BEARER.value: BearerAuthStrategy,
        AuthType.API_KEY.value: APIKeyAuthStrategy,
        AuthType.CUSTOM_HEADER.value: CustomHeaderAuthStrategy,
        AuthType.OAUTH2_CLIENT_CREDENTIALS.value: OAuth2ClientCredentialsAuthStrategy,
    }

    @classmethod
    def create_auth_strategy(cls, auth_type: str | AuthType) -> IMCPAuthStrategy:
        """Create an authentication strategy for the given auth type.

        Args:
            auth_type: String identifier for auth type (e.g., "basic", "bearer")

        Returns:
            Authentication strategy instance

        Raises:
            ValueError: If auth type is not supported
        """
        # Support both enum and string inputs
        normalized = (
            auth_type.value if isinstance(auth_type, AuthType) else str(auth_type)
        ).lower()
        strategy_class = cls._strategy_map.get(normalized)

        if not strategy_class:
            supported = list(cls._strategy_map.keys())
            raise ValueError(
                f"Unsupported auth type: {auth_type}. Supported: {supported}"
            )

        return strategy_class()

    @classmethod
    def get_supported_auth_types(cls) -> list[str]:
        """Get list of supported authentication types."""
        return list(cls._strategy_map.keys())

def main():
    # 1. Resolve Auth
    auth_type = "basic"
    auth_strategy = AuthFactory.create_auth_strategy(auth_type)
    httpx_auth = auth_strategy.create_auth(creds)
    headers = auth_strategy.get_auth_headers(creds)
```
