## UML Diagrams

Unified Modeling Language (UML) diagrams are standardized visual, object-oriented modeling tools used to design, document, and visualize software systems and business processes

UML is a standardized modeling language (approved by ISO) that provides a common way for teams to visualize and communicate system design.

There are two types of UML Diagram:
1. Structural
2. Behavioural

### Structural Diagram: 
Focus on what is in the system, representing static elements like classes, objects, packages, components and deployment nodes.
1. Class Diagram: Shows classes, attributes, methods and relationships.
2. Components Diagram: Visualize organization and dependencies among software components.
3. Deployment Diagram: Maps software artifacts to hardware components.

### Behavioral Diagrams:
Focus on how the system acts, representing dynamic interactions and states.
1. Sequence Diagram: Shows interactions between objects over time.
2. Use case Diagram: Describes system functionality from user's perspective.
3. Activity Diagram: Illustrates the step-by-step workflow of systems.


### Class Diagram
class diagram is the building block of all object oriented software systems. We use class diagrams to depict the static structure of a system by showing system's classes, their methods and attributes.

Class Diagram has 2 representations:
1. Class Structure



2. Associations


#### Class Associations:
Inheritance: is a relationship (->)

Scenario: 
1. Bus is a vehicle
2. Car is a Vehicle
3. Truck is a Vehicle



#### Object Associations

Simple Association: has-a relationship. It is a weaker relationship.

Scenario: Arjun has a house, Person has a house



Aggregation: is a parent object where it has smaller objects, where each object can exist independently.
It has sronger relationship compared to simple association.

Scenario: Room has a sofa, chair, bed. Team has employees.

Composition: Where each object cannot exist independently. It has a strong relationship compared to aggregation.

Scenario: Chair has a arms, wheels, arms.








### Behavioural Diagram

#### Sequence Diagram
A sequence diagram simply depicts interaction between objects in a sequential order i.e. the order in which these interactions take place.

Follow below steps for Sequence diagram:
1. Identify Objects / class
2. Lifeline: It will tell upto what time object will exist.
3. activation Bar: Tells upto what time object is active.
4. Messages can be: Sync vs Async, Create Message vs Destroy Message, Lost Message vs Found Message.
